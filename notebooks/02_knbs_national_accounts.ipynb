```python
import pandas as pd

# KNBS source file
file_path = "../raw/NA-Statistical-Abstract_2025-1.xlsx"

# Read Table 2.8 - 2.9 exactly as it appears in the workbook
raw_table = pd.read_excel(
    file_path,
    sheet_name="Table 2.8 - 2.9",
    header=None
)

print("Table loaded successfully.")
print("Shape:", raw_table.shape)

# Find the row containing GDP at market prices
target = "gross domestic product at market prices"

matching_rows = []

for row_number in range(len(raw_table)):
    row_text = " ".join(
        str(value).lower()
        for value in raw_table.iloc[row_number].tolist()
        if pd.notna(value)
    )

    if target in row_text:
        matching_rows.append(row_number)

print("\nRows containing GDP at market prices:", matching_rows)
```

```python
import re
import pandas as pd

# Row containing the years
year_row = raw_table.iloc[3, 1:11]

# Row containing GDP at market prices
gdp_row = raw_table.iloc[9, 1:11]

# Clean the years
years = []

for value in year_row:
    text = str(value)
    match = re.search(r"\d{4}", text)

    if match:
        years.append(int(match.group()))
    else:
        years.append(None)

# Convert GDP values to numbers
gdp_values = pd.to_numeric(gdp_row, errors="coerce")

# Create clean DataFrame
gdp_df = pd.DataFrame({
    "year": years,
    "gdp_current_ksh_million": gdp_values.values
})

print("Clean GDP dataset:")
display(gdp_df)
```

```python
# Search the KNBS workbook for GDP growth-related entries

search_terms = ["gdp growth", "growth rate", "economic growth"]

matches = []

for sheet in excel_file.sheet_names:
    df = pd.read_excel(
        file_path,
        sheet_name=sheet,
        header=None
    )

    for row_number in range(len(df)):
        row_text = " ".join(
            str(value).lower()
            for value in df.iloc[row_number].tolist()
            if pd.notna(value)
        )

        if any(term in row_text for term in search_terms):
            matches.append({
                "sheet": sheet,
                "row": row_number,
                "text": row_text[:300]
            })

print("Potential GDP growth entries found:")
display(pd.DataFrame(matches))
```

```python
from scipy import stats
import pandas as pd

# Make sure Final_data is loaded
vars = ["gdp_current_ksh_million", "gdp_growth_percent", 
        "gfcf_ksh_million", "unemployment_rate_percent", "inflation_rate_percent"]

print("Pearson Correlations with p-values\n" + "="*50)

for i in range(len(vars)):
    for j in range(i+1, len(vars)):
        r, p = stats.pearsonr(Final_data[vars[i]], Final_data[vars[j]])
        sig = "***" if p < 0.01 else "**" if p < 0.05 else "*" if p < 0.1 else ""
        print(f"{vars[i]}  ↔  {vars[j]}")
        print(f"   r = {r:.3f}   |   p = {p:.4f}  {sig}")
        print()
```

```python
import statsmodels.api as sm
import numpy as np

# Prepare data
df = Final_data.copy()
df["log_gdp"] = np.log(df["gdp_current_ksh_million"])

# Model A: Unemployment ~ GDP growth + Inflation
X_a = sm.add_constant(df[["gdp_growth_percent", "inflation_rate_percent"]])
model_a = sm.OLS(df["unemployment_rate_percent"], X_a).fit()

print("="*60)
print("MODEL A: Unemployment ~ GDP Growth + Inflation")
print("="*60)
print(model_a.summary())

# Model B: Unemployment ~ log(GDP)
X_b = sm.add_constant(df["log_gdp"])
model_b = sm.OLS(df["unemployment_rate_percent"], X_b).fit()

print("\n" + "="*60)
print("MODEL B: Unemployment ~ log(GDP)")
print("="*60)
print(model_b.summary())
```
